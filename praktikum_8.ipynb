```python
#Activity 6F

scores = [80, 65, 90, 75, 70]

total = 0                    # accumulator: starts at zero
count_pass = 0               # a counter is an accumulator that adds 1
for score in scores:
    total = total + score    # add each element
    if score >= 75:
        count_pass = count_pass + 1
    print("running total:", total)

average = total / len(scores)
print("Total   :", total)
print("Average :", average)
print("Scores >= 75:", count_pass, "of", len(scores))
```

```python
#Activity 6G

# Data suhu selama 7 hari
temps = [25, 26, 27, 28, 29, 30, 31]

# Menghitung moving average 5 hari
averages = []

for i in range(4, len(temps)):
    window = temps[i - 4 : i + 1]
    average = sum(window) / 5
    averages.append(average)

# Menampilkan hasil
print("Data suhu:", temps)
print("Moving average 5 hari:", averages)
print("Jumlah rata-rata:", len(averages))
```

```python
#Activity 6 H

# Data operasional selama 30 hari
energy = [
    100, 105, 110, 108, 115, 112, 100, 105, 110, 108,
    115, 112, 100, 105, 110, 108, 60, 112, 100, 105,
    110, 108, 115, 112, 100, 105, 110, 108, 115, 112
]

run_hours = [
    24, 24, 24, 24, 24, 22, 20, 24, 24, 24,
    24, 22, 20, 24, 24, 24, 12, 24, 20, 24,
    24, 24, 24, 22, 20, 24, 24, 24, 24, 22
]

# Mengambil energi hari ke-16 dan ke-17
energy_before = energy[15]
energy_day17 = energy[16]

# Menghitung persentase penurunan
decrease = ((energy_before - energy_day17) / energy_before) * 100

print("Energi hari ke-16:", energy_before)
print("Energi hari ke-17:", energy_day17)
print("Jam operasi hari ke-17:", run_hours[16])
print("Persentase penurunan:", round(decrease, 2), "%")

# Memeriksa apakah penurunan lebih dari 40%
if decrease > 40:
    print("Energi hari ke-17 mengalami penurunan lebih dari 40%.")
else:
    print("Penurunan energi tidak lebih dari 40%.")
```

```python
#Activity 6I

# Membuat list suhu selama 7 hari
temps = [28, 30, 27, 32, 29, 31, 26]

print("Data suhu selama 7 hari:", temps)

# 1. Menampilkan 3 hari pertama dan 2 hari terakhir
print("\n3 hari pertama:", temps[:3])
print("2 hari terakhir:", temps[-2:])

# 2. Menambahkan satu data menggunakan append()
temps.append(33)
print("\nSetelah append:", temps)

# Menghapus data terakhir menggunakan pop()
temps.pop()
print("Setelah pop:", temps)

# 3. Mengurutkan data tanpa mengubah list asli
sorted_temps = sorted(temps)
print("\nData suhu setelah diurutkan:", sorted_temps)
print("Data suhu asli:", temps)

# 4. Menghitung rata-rata suhu
average = sum(temps) / len(temps)
print("\nRata-rata suhu:", round(average, 2))

# Mencari suhu tertinggi dan hari terjadinya
highest = max(temps)
day = temps.index(highest) + 1

print("Suhu tertinggi:", highest)
print("Suhu tertinggi terjadi pada hari ke-", day)

# 5. Mencari suhu yang berada di atas rata-rata
above_average = []

for temp in temps:
    if temp > average:
        above_average.append(temp)
```

```python
#Lima latihan kecil

# Membuat list skor mahasiswa
skor = [80, 65, 90, 75, 70, 85]

# 1. Mencetak skor pertama dan terakhir
print("Skor pertama:", skor[0])
print("Skor terakhir:", skor[-1])

# 2. Menghitung rata-rata skor
rata_rata = sum(skor) / len(skor)
print("\nRata-rata skor:", rata_rata)

# 3. Menghitung jumlah skor di atas rata-rata
jumlah_di_atas = 0

for nilai in skor:
    if nilai > rata_rata:
        jumlah_di_atas += 1

print("Jumlah skor di atas rata-rata:", jumlah_di_atas)

# 4. Mencari posisi skor terendah sebagai nomor mahasiswa
skor_terendah = min(skor)
posisi = skor.index(skor_terendah) + 1

print("Skor terendah:", skor_terendah)
print("Nomor mahasiswa:", posisi)

# 5. Membuat list baru dengan setiap skor ditambah 5
skor_baru = []

for nilai in skor:
    skor_baru.append(nilai + 5)

print("Skor setelah ditambah 5:", skor_baru)
```

```python
# LAPORAN PRODUKSI ENERGI BULANAN

import pandas as pd

# 1. Membuat data produksi energi selama 30 hari
data = {
    "tanggal": pd.date_range(start="2026-01-01", periods=30),
    "energi": [
        120, 125, 118, 130, 115, 122, 128,
        110, 115, 120, 125, 118, 130, 122,
        105, 110, 100, 115, 120, 125, 118,
        130, 125, 110, 115, 120, 105, 100,
        125, 130
    ]
}

# Membuat DataFrame
df = pd.DataFrame(data)

# 2. Menghitung total dan rata-rata energi
total_energi = df["energi"].sum()
rata_rata = df["energi"].mean()

print("=== LAPORAN PRODUKSI BULANAN ===")
print("Total energi:", total_energi)
print("Rata-rata energi:", round(rata_rata, 2))

# 3. Mencari 3 hari tertinggi dan terendah
print("\n=== 3 HARI PRODUKSI TERTINGGI ===")
tertinggi = df.nlargest(3, "energi")
print(tertinggi[["tanggal", "energi"]].to_string(index=False))

print("\n=== 3 HARI PRODUKSI TERENDAH ===")
terendah = df.nsmallest(3, "energi")
print(terendah[["tanggal", "energi"]].to_string(index=False))

# 4. Menghitung jumlah hari di bawah target
target = float(input("\nMasukkan target energi harian: "))

jumlah_di_bawah = (df["energi"] < target).sum()

print("\n=== PRODUKSI DI BAWAH TARGET ===")
print("Target energi:", target)
print("Jumlah hari di bawah target:", jumlah_di_bawah)

# 5. Menghitung moving average 7 hari
df["rata_rata_7_hari"] = df["energi"].rolling(window=7).mean()

batas = rata_rata * 0.97

print("\n=== MOVING AVERAGE 7 HARI ===")
print("Batas 97% rata-rata bulanan:", round(batas, 2))

hasil = df[df["rata_rata_7_hari"] < batas]

if len(hasil) > 0:
    print("Hari dengan rata-rata bergerak di bawah batas:")
    print(hasil[["tanggal", "rata_rata_7_hari"]].to_string(index=False))
else:
    print("Tidak ada hari di bawah batas.")

# 6. Menghitung total energi setiap minggu
df["kelompok_minggu"] = df.index // 7 + 1

print("\n=== TOTAL ENERGI PER MINGGU ===")

for minggu, kelompok in df.groupby("kelompok_minggu"):
    total_minggu = kelompok["energi"].sum()

    if minggu <= 4:
        print(f"Minggu ke-{minggu}: {total_minggu}")
    else:
        print(f"Sisa hari: {total_minggu}")

# 7. Menampilkan seluruh data
print("\n=== DATA PRODUKSI 30 HARI ===")
print(df.to_string(index=False))
```
