```python
#SOAL NO 5 Katup pengaman


tekanan_uap = float(input("Masukkan tekanan uap (psi): "))
status_manual_override = False


if tekanan_uap > 150 and status_manual_override == False:
    status_katup = "Katup Terbuka"
elif status_manual_override == True:
    status_katup = "Katup Mode Manual"
else:
    status_katup = "Katup Tertutup"

# Menampilkan hasil
print(f"Status katup = {status_katup}")
```

```python
#SOAL NO 7 Pemanasan mencapai target

suhu_saat_ini = 60
while suhu_saat_ini < 85:
    suhu_saat_ini += 5

    print(f"Suhu saat ini: {suhu_saat_ini} C")
    if suhu_saat_ini == 80:
      print("WARNING: Mendekati target!")

    print("Target 85 C tercapai.")

```

```python
#SOAL NO 9 Pembacaan vibrasi dengan sensor rusak

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0
for nomor, nilai in enumerate(data, start=1):
    try:
        vibrasi = float(nilai)

        if vibrasi > 2.0:
            print(f"Warning: Unit {nomor} vibrasi tinggi!")

        if vibrasi > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

    except ValueError:
        print(f"Unit {nomor}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

# Menampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")

```
