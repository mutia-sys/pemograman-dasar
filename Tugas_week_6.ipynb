```python
# Soal 1 - kalkulator faktorial 1 sampai 5
n = int(input("Masukan bilangan bulat (1-5): "))

#Validasi input
if n < 1 or n > 5:
  print("Input tidak valid. Masukan bilangan antara 1 sampai 5.")

#Perhitungan faktorial
elif n == 1:
  print("1! = 1")

elif n == 2:
  hasil = 2 * 1
  print("2! =", hasil)

elif n == 3:
  hasil = 3 * 2 * 1
  print("3! =", hasil)

elif n == 4:
  hasil = 4 * 3 * 2 * 1
  print("4! =", hasil)

else:
  hasil = 5 * 4 * 3 * 2 * 1
  print("5! =", hasil)
```

```python
# Soal 2 - Klasifikasi daya
faktor_daya = float(input("Masukan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukan efisiensi distribusi (%): "))

#Menghitung skor klualitas daya
skor = faktor_daya * efisiensi

print(f"Skor kualitas daya: {skor:.2f}")

#Menentukan klasifikasi kualitas daya
if skor >= 85 and faktor_daya >= 0.90:
  klasifikasi = "Prima - sistem dalam kondisi optimal"

elif skor >= 70 and faktor_daya >= 0.80:
  klasifikasi = "BAIK - sistem memadai, pantau rutin"

elif skor >= 50 or faktor_daya >= 0.75:
  klasifikasi = "CUKUP - perlu ruang perbaikan"

else:
  klasifikasi = "BURUK - perlu perbaikan segera"

print(f"Faktor daya {faktor_daya}, efisiensi {efisiensi}% → klasifikasi: {klasifikasi}")
```

```python
# Saol 3 - Rekomendasi mode operasi PLTS

iradiasi = float(input("Masukan iradiasi matahari (W/m²): "))
soc = float(input("Masukan SoC baterai (%): "))

# menghitung estimasi day apanel
daya_panel = iradiasi * 0.18 * 10

print(f"estimasi daya panel: {daya_panel:.1f} W")

#Menentukan mode operasi PLTS
if iradiasi > 600 and soc < 80:
  mode = "CHARGING —  ekspor kelebihan daya ke jaringan"

elif iradiasi > 600 and soc >= 80:
  mode = "GRID-EXPORT — operasi mandiri"

elif iradiasi >= 200 and soc >= 50:
  mode = "BATTERY-ONLY — andalkan batrai"

else:
  mode = "EMERGENCY — kurangi beban segera"

print(f"iradiasi {iradiasi} W/m², SoC {soc}% → Mode: {mode}")
```
