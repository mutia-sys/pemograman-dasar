```python
# 1. Nama : Mutia salsabila p
# 2. Nim : 4232501009

# SOAl 1 (Belanja buku)

jumlah_buku = 7
harga_per_buku = 18500
diskon = 10 / 100
uang_budi = 150000

#total belanja
total_belanja = jumlah_buku * harga_per_buku

#diskon
total_diskon = total_belanja * diskon

#yang harus di bayar
jumlah_dibayar = total_belanja - total_diskon

#uang kembalian
uang_kembalian = uang_budi - jumlah_dibayar

#iuran per orang
jumlah_orang = 4
iuran_per_orang = jumlah_dibayar // jumlah_orang
sisa = jumlah_dibayar % jumlah_orang
uang_cukup = uang_budi >= jumlah_dibayar

# tampilkan hasil perhitungan
print("total belanja :", total_belanja)
print("total diskon :", total_diskon)
print("jumlah yang dibayar :", jumlah_dibayar)
print("uang kembalian :", uang_kembalian)
print("iuran per orang :", iuran_per_orang)
print("sisa pembagian :", sisa)
print("uang cukup :", uang_cukup)
```

```python
#SOAL 2

#PLTS atap rumah
jumlah_panel = 12
daya = 450
penyinaran = 5
efesiensi = 0.8
kebutuhan_listrik_rumah = 25
tarif_listrik = 1444.70

#energi ynag di hasilkan per hari (kWh)
total_daya = jumlah_panel * daya
energi_harian = (total_daya * penyinaran * efesiensi) / 1000

#energi per bulan
energi_bulan =  energi_harian * 30
nilai_listrik = energi_bulan * tarif_listrik

#kekurangan energi per hari terhadap kebutuhan
kekurangan_energi = kebutuhan_listrik_rumah - energi_harian

#apakah PLTS sudah mencukupi kebutuhan rumah
plts_mencukupi = energi_harian >= kebutuhan_listrik_rumah

#tampilkan hasil
print("total daya :", total_daya)
print("energi harian :", energi_harian)
print("energi bulan :", energi_bulan)
print("nilai listrik :", nilai_listrik)
print("kekurangan energi :", kekurangan_energi)
print("plts mencukupi :", plts_mencukupi)
```

```python
#SOAL 3

#generator disel
daya_generator = 200
jam_operasi = 8
konsumsi_bbm = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

#energi yang dibangkitkan perhari
energi_harian = daya_generator * jam_operasi

#kebutuhan solar
solar_harian = energi_harian * konsumsi_bbm
biaya_harian = solar_harian * harga_solar

#berapa hari penuh tangkicukup, berapa liter solar yang tersisa
hari_penuh = int(kapasitas_tangki // solar_harian)
sisa_solar = kapasitas_tangki % solar_harian

#rugi daya pada kabel
rugi_daya = arus**2 * hambatan

#mengecek apakah rugi kabel tergolong tinggi
rugi_tinggi = (rugi_daya > 150) and (arus >= 20)

#tampilkan hasil
print("energi harian :", energi_harian)
print("solar harian :", solar_harian)
print("biaya harian :", biaya_harian)
print("hari penuh :", hari_penuh)
print("sisa solar :", sisa_solar)
print("rugi daya :", rugi_daya)
print("rugi tinggi :", rugi_tinggi)
```
