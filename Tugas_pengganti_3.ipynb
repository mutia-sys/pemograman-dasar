1). Urutan prioritas operator

Menurut dokumentasi resmi Python, setiap operator memiliki prioritas (precedence). Operator dengan prioritas lebih tinggi akan dikerjakan terlebih dahulu. Pada operasi aritmatika, perkalian * memiliki prioritas lebih tinggi daripada penjumlahan +.

Contoh
10 + 5 * 2
python tidak langsung menghitung 10 + 5 terlebih dahulu. Karena * memiliki prioritas lebih tinggi daripada +, maka Python menghitung:
5 * 2 = 10
hasil dijumlahkan
10 + 10 = 20
Jadi, hasil akhirnya adalah 20, bukan 30.

Kalau ingin penjumlahan dilakukan terlebih dahulu, harus menggunakan tanda kurung:
(10 + 5) * 2
Hasilnya menjadi:
15 * 2 = 30

jadi operator precedence digunakan Python untuk menentukan operator mana yang dikerjakan terlebih dahulu ketika terdapat beberapa operator dalam satu ekspresi.

2. Perbedaan operator rasional dan operator logika

a. Operator relasional digunakan untuk membandingkan dua nilai. Hasilnya berupa True atau False. Contohnya:
> → lebih besar dari
< → lebih kecil dari
== → sama dengan
>= → lebih besar atau sama dengan
<= → lebih kecil atau sama dengan
!= → tidak sama dengan

b. Operator logika
digunakan untuk menggabungkan atau membalik hasil kondisi. Contohnya:
and → kedua kondisi harus benar
or → salah satu kondisi benar
not → membalik hasil kondisi

contoh
Misalnya, menentukan apakah seseorang boleh mengikuti ujian. Syaratnya adalah nilai minimal 75 dan kehadiran minimal 80%.

nilai = 80
kehadiran = 85

boleh_ujian = nilai >= 75 and kehadiran >= 80

print(boleh_ujian)

hasil
True

penjelasan:
nilai >= 75 → True → operator relasional
kehadiran >= 80 → True → operator relasional
and menggabungkan kedua kondisi tersebut → operator logika
Karena kedua kondisi benar, hasil akhirnya adalah True.

Jadi, operator relasional menentukan apakah suatu kondisi benar atau salah, sedangkan operator logika menggabungkan beberapa kondisi untuk menghasilkan suatu keputusan.
